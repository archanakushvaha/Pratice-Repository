```python
import matplotlib.pyplot as plt
```

```python
# 1 Traffic Sources

month = ["JAN" , "FEB" , "MAR" , "APR" , "MAY" , "JUN"]

google_search = [25 , 34 , 42 , 56 , 60 , 43]
social_media = [54 , 45 , 55 , 60 , 50 , 40]
direct_visitors = [35 , 60 , 50 , 20 , 45 , 60]

plt.figure(figsize=(10 , 5))

plt.stackplot(
    month , 
    google_search ,
    social_media , 
    direct_visitors,
    
    labels=[
        "google_search",
        "social_media",
        "direct_visitors"
    ],
    
    colors=[
        "pink",
        "blue",
        "green"
    ]
)

plt.title("traffic sources contribute")

plt.xlabel(month)

plt.ylabel("traffic sources")

plt.legend()

plt.show()
```
