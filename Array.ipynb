Basic filtering

```python
ages = np.array([25 , 30 , 18 , 35 , 22 , 40 , 17])

adults = ages[ages >= 18]

print(adults)
```

```python
adults_age = ages[(ages > 20) & (ages < 35)]

print(adults_age)
```

```python
names = np.array(["Rahul" , "Amit",  "Alice" , "Pooja" , "Carlie" , "Bob"])

selected_name = names[np.char.startswith(names , "A") | np.char.startswith(names , "P")]

print(selected_name)
```

Basic Masking for Selection

```python
temp = np.array([28 , 32 , 35 , 30 , 22 , 38])

is_hot = temp > 30

hot_temp = temp[is_hot]

print(temp)
print(is_hot)
print(hot_temp)
```

Masking Modification

```python
temp[is_hot] = 31

print(temp)
```

np.where()

```python
scores = np.array([68 , 81 , 78 , 90 , 55 , 78])

# If score >= 75 -> Pass , otherwise Fail

grades = np.where(scores >= 75 , "Pass" , "")

print(scores)
print(grades)
```

Reshaping 1D and 2D

```python
num_1d = np.array([1 , 2 , 3 , 4])

num_2d = num_1d.reshape(1 , 4)

print(num_1d)

print(num_2d)
```

Broadcasting to an array

```python
prices = np.array([10 , 20 , 30])

# add 5 to every element

new_prices = prices + 5

print(prices)

print(new_prices)
```

Broadcasting 1D to a 2D

```python
sales = np.array([[10 , 20] , [15 , 25] , [12 , 18]])

add_value = np.array([2 , 3 , 1])

print(sales)

print(add_value)
```

```python
bonues_col =add_value.reshape(3 , 1)

sales_with_bonues = sales + bonues_col

print(bonues_col)
```
