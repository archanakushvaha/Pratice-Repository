#### Summarize , Visulization , Merge and Joine Data

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'Name' : ["Riya" , "Priya" , "Rahul" , "Rakesh" , "Rita"],
    'Age' : [21 , 19 , 20 , 25 , 23],
    'Gender' : ['Female' , 'Female' , 'Male' , 'Male' , 'Female'],
    'Class' : [1 , 2 , 1 , 1 ,3],
    'Fare' : [500 , 800 , 200 , 450 , 550]
})

print(df)
print(df.dtypes)
```

```python
sorted_df = df.sort_values("Age")

print(sorted_df)
```

```python
filter_df = df[df['Age'] <= 20]

print(filter_df)
```
